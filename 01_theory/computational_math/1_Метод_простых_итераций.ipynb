```python
import math

def simple_iteration(phi, x0, eps=1e-6, max_iter=50):
    x = x0
    
    for i in range(max_iter):
        x_next = phi(x)
        
        if abs(x_next - x) < eps:
            print(f"Сошлось за {i+1} итераций")
            return x_next
        
        x = x_next
    
    print("Достигнуто максимальное число итераций")
    return x



func_str = input("Введите функцию phi(x), например, math.sqrt(1/(3*math.atan(x))): ")
x0 = float(input("Введите начальное приближение x0: "))


phi = eval(f"lambda x: {func_str}")
# phi = lambda x: math.sqrt(1/(3*math.atan(x)))

root = simple_iteration(phi, x0)
print(f"Приближённый корень: {root:.6f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import math

def f1(x):
    return np.arctan(x)

def f2(x):
    return 1 / (3 * x**2)

def phi(x):
    return math.sqrt(1 / (3 * math.atan(x)))

x0 = 0.7
eps = 1e-8

max_iter = 50
i=0
x = x0
while True:
    x_next = phi(x)
    if abs(x_next - x) < eps or i == max_iter:
        break
    x = x_next
    i+=1

root = x_next
y_root = math.atan(root)

x_vals = np.linspace(0.1, 2, 1000)

plt.figure()
plt.plot(x_vals, f1(x_vals))
plt.plot(x_vals, f2(x_vals))
plt.scatter(root, y_root)
plt.title("Пересечение функций arctan(x) и 1/(3x^2)")
plt.xlabel("x")
plt.ylabel("y")
plt.show()

print(f"Точка пересечения: x ≈ {root:.6f}, y ≈ {y_root:.6f}")
```
