# Анализ набора данных о диабете

## Описание
Данные взяты из [sklearn.datasets.load_diabetes](https://scikit-learn.org/stable/modules/generated/sklearn.datasets.load_diabetes.html).  
Набор содержит **442 наблюдения** за пациентами с сахарным диабетом. Для каждого пациента доступны:

- **10 исходных признаков**:
  - возраст (`age`)
  - пол (`sex`)
  - индекс массы тела (`bmi`)
  - среднее артериальное давление (`bp`)
  - шесть показателей сыворотки крови (`s1`–`s6`)
- **Целевая переменная**:
  - количественная мера прогрессирования заболевания через год (`target`)

Все признаки были предварительно стандартизированы (центрированы и нормированы).
