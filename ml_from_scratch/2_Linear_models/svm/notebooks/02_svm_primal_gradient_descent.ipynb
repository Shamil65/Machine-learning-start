```python
import sys
from pathlib import Path

# ноутбук: .../svm/notebooks/02_primal_svm_gradient_descent.ipynb
# svm.py:  .../svm/svm.py  -> на один уровень выше текущей папки
sys.path.append(str(Path.cwd().parent))

%load_ext autoreload
%autoreload 2

from primal_svm import MySVM
```

```python
import pandas as pd
from sklearn.datasets import make_classification

X, y = make_classification(
    n_samples=200, n_features=2, n_informative=2, n_redundant=0,
    n_clusters_per_class=1, class_sep=2.5, random_state=42
)
X_df = pd.DataFrame(X, columns=['x1', 'x2'])
y_series = pd.Series(y)

model = MySVM(n_iter=50, learning_rate=0.01, C=1)
model.fit(X_df, y_series, verbose=10)

preds = model.predict(X_df)
weights, b = model.get_coef()
print(weights, b)
```
